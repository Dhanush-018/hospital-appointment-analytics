```python
%sql
SELECT COUNT(*) AS previous_silver_count
FROM hospital_appointments.silver.silver_data;
```

```python
%sql
WITH batch2_valid AS (
    SELECT appointment_id
    FROM hospital_appointments.silver.stg_batch2_valid
),
counts AS (
    SELECT
        COUNT(*) AS accepted_count,
        SUM(CASE WHEN s.appointment_id IS NULL THEN 1 ELSE 0 END) AS insert_count,
        SUM(CASE WHEN s.appointment_id IS NOT NULL THEN 1 ELSE 0 END) AS update_count
    FROM batch2_valid v
    LEFT JOIN hospital_appointments.silver.silver_data s
        ON v.appointment_id = s.appointment_id
)
SELECT
    88 AS previous_silver_count,
    accepted_count,
    insert_count,
    update_count,
    88 + insert_count AS expected_silver_count
FROM counts;
```
