```python
file_path='/Volumes/hospital_appointments/source/input_raw_files/Hospital_Appointments_Intial.csv'
```

```python
hospital_df=(
    spark.read.format("csv")
    .option("header","true")
    .option("inferSchema","true").load(file_path)
)
display(hospital_df)
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from delta.tables import DeltaTable
```

```python
hospital_df.printSchema()
```

```python
hospital_df=(
   hospital_df
    .withColumn("source_file",lit(file_path))
    .withColumn("batch_id",lit("BATCH_01"))
    .withColumn("ingestion_timestamp",current_timestamp())
)
display(hospital_df)
```

```python
total_records=hospital_df.count()
print(f"Total records in the table: " ,total_records)
```

```python
hospital_df.write.mode("overwrite").format("delta").saveAsTable("hospital_appointments.bronze.batch1_raw_data")
```
